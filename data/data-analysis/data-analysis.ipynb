## Loading metrics and ranks

```python
from scipy.stats import rankdata
```

```python
def get_updated_metrics(path, suffix='_sd', update_nli = True, save_file = False):
    with open(path+f'all_metric_results{suffix}.json', 'r', encoding='utf-8') as file:
        spandist_metrics_dict = json.load(file)
    with open(path+'nli_exp_full_comp.json', 'r', encoding='utf-8') as file:
        nli_metrics_dict = json.load(file)

    new_metrics_dict = spandist_metrics_dict.copy()

    if update_nli:
        for key,val in new_metrics_dict.items():
            for inner_key,inner_val in val.items():
                inner_val['metrics'].update(
                    nli_metrics_dict[key][inner_key]['metrics']
                )
    
    if save_file:
        with open(path+"new_metric_results.json", "w") as json_file:
            json.dump(new_metrics_dict, json_file, indent=4)

    return new_metrics_dict

# get_updated_metrics('b/', save_file = True)
# get_updated_metrics('c/', save_file = True)

# new_metrics_b = get_updated_metrics('b/', update_nli=False)
# new_metrics_c = get_updated_metrics('c/', update_nli=False)

# new_metrics_b = get_updated_metrics('b/', suffix='_ex', update_nli=False)
# new_metrics_c = get_updated_metrics('c/', suffix='_ex', update_nli=False)

new_metrics_b = get_updated_metrics('b/')
new_metrics_c = get_updated_metrics('c/')
```

```python
def get_rank_correlation(metrics_dict, hm_rank_idx, hm_rank_1, hm_rank_2, llm_rank_first, llm_rank_improv, label=None):
    outer_idx = 0
    metrics_mtx = np.zeros([486,8,3])
    for key,val in metrics_dict.items():
        inner_idx = 0
        for inner_key in ['zero', 'sft', 'berzak']:
            metrics_mtx[outer_idx,:,inner_idx] = [v for v in val[inner_key]['metrics'].values()]
            inner_idx += 1
        outer_idx += 1
    
    metrics_rank_mtx = rankdata(-metrics_mtx, method = 'min', axis = 2)[hm_rank_idx,:,:]
    print(metrics_rank_mtx.shape)

    metrics_label = [k for k in metrics_dict['1']['zero']['metrics'].keys()]

    metrics_mean_df = pd.DataFrame(
        metrics_mtx.mean(axis=0).T,
        index = ['zero', 'sft', 'berzak'],
        columns = [k for k in metrics_dict['1']['zero']['metrics'].keys()]
        ).round(4)

    metrics_rank_dict = dict()
    for i in range(len(metrics_label)):
        metrics_rank_dict[metrics_label[i]] = metrics_rank_mtx[:,i,:].tolist()

    cor_dict = dict()
    for l in metrics_label:
        taus = [
            kendalltau(metrics_rank_dict[l], hm_rank_1)[0],
            kendalltau(metrics_rank_dict[l], hm_rank_2)[0],
            kendalltau(metrics_rank_dict[l], llm_rank_first)[0],
            kendalltau(metrics_rank_dict[l], llm_rank_improv)[0],
        ]
        cor_dict[l] = taus
    
    cor_df = pd.DataFrame(
        cor_dict, 
        index=['hm1', 'hm2', 'gpt4o (first)', 'gpt4o (+)']) \
        .round(4)

    metrics_mean_df = metrics_mean_df[['bleu', 'rouge', 'bleurt_s', 'bart_s', 'bert_s', 'entail', 'neutral', 'contra']]
    cor_df = cor_df[['bleu', 'rouge', 'bleurt_s', 'bart_s', 'bert_s', 'entail', 'neutral', 'contra']]

    if label:
        metrics_mean_df['label'] = label
        cor_df['label'] = label

    return metrics_mtx, metrics_mean_df, cor_df

b_mtx, b_mean_df, b_cor_df = get_rank_correlation(
    new_metrics_b, 
    hm_rank_idx_b, 
    hm_rank_1_b, hm_rank_2_b, 
    llm_rank_first_b,
    llm_rank_improv_b,
    label='b'
    )
c_mtx, c_mean_df, c_cor_df = get_rank_correlation(
    new_metrics_c, 
    hm_rank_idx_c, 
    hm_rank_1_c, hm_rank_2_c, 
    llm_rank_first_c,
    llm_rank_improv_c,
    label='c'
    )
```

```python
all_mean_df = pd.concat([
    b_mean_df, 
    c_mean_df
]).reset_index()

all_cor_df = pd.concat([
    b_cor_df, 
    c_cor_df
]).reset_index()
```

```python
metrics = ['bleu', 'rouge', 'bleurt_s', 'bart_s', 'bert_s', 'entail', 'neutral', 'contra']
title_mapping = dict({
    'bleu': 'BLEU',
    'rouge': 'ROUGE',
    'bleurt_s': 'BLEURT',
    'bart_s': 'BARTSc.',
    'bert_s': 'BERTSc.',
    'entail': 'Ent.',
    'neutral': 'Neu.',
    'contra': 'Con.'
})

metric_data = all_mean_df
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for i, metric in enumerate(metrics):
    metric_data = all_mean_df[metric]
    
    categories = ['zero', 'sft', 'berzak']
    group1_values = [metric_data.iloc[0], metric_data.iloc[1], metric_data.iloc[2]]
    group2_values = [metric_data.iloc[3], metric_data.iloc[4], metric_data.iloc[5]]
    
    x = np.arange(len(categories))
    
    # 绘制折线图（虚线连接点）
    line1 = axes[i].plot(x, group1_values, 'o--', linewidth=2, markersize=10, label='B', alpha=0.8)
    line2 = axes[i].plot(x, group2_values, 's--', linewidth=2, markersize=10, label='C', alpha=0.8)
    
    y_min, y_max = axes[i].get_ylim()
    y_range = y_max - y_min
    
    # 扩展y轴范围以容纳标注
    padding = y_range * 0.15  # 15%的额外空间
    axes[i].set_ylim(y_min - padding, y_max + padding)
    
    # 添加数值标注
    for j, (val1, val2) in enumerate(zip(group1_values, group2_values)):
        # 根据数值大小调整标注位置
        offset = 0.02 * (max(max(group1_values), max(group2_values)) - min(min(group1_values), min(group2_values)))
        
        axes[i].text(x[j], val1, f'{val1:.2f}', 
                    ha='center', 
                    va='top' if val1<val2 else 'bottom', 
                    fontsize=22)
        
        axes[i].text(x[j], val2, f'{val2:.2f}', 
                    ha='center', 
                    va='top' if val2<val1 else 'bottom',
                    fontsize=22)
    
    axes[i].set_title(title_mapping[metric], fontsize=22)
    axes[i].set_xlim(-.5, 2.5)
    axes[i].set_xticks(x)
    axes[i].set_xticklabels(categories, fontsize=20)
    axes[i].grid(True, alpha=0.7, linestyle=':')
    
    if i == 0:
        axes[i].legend(fontsize=16)

plt.tight_layout()
plt.show()
```
